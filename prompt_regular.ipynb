# Regular prompt — P1-VL-30B-A3B (Q4_K_M)

The fixed prompt lives in `prompts/prompt_regular.md` and is applied to all three sets. Question files: `regular_questions.json` (regular prompt only), `obvious_questions.json` and `non_obvious_questions.json` (both prompts).

Answers are saved as `outputs/Q<n>_<set>_regular_prompt.md`, e.g. `outputs/Q1_obvious_regular_prompt.md`, plus a line in `outputs/log.jsonl`. Type a question in the box (pick its set) and press **Run**, or use the batch cell at the bottom.

```python
# One-time setup is in README.md (llama-cpp-python needs a CUDA build).
# Quick check that the GPU build is installed:
import llama_cpp; print(llama_cpp.__version__, 'GPU:', llama_cpp.llama_supports_gpu_offload())
```

```python
PROMPT_NAME = "prompt_regular"   # the prompt text is in prompts/prompt_regular.md
SEED = None          # set an int (e.g. 0) for reproducible sampling
GEN_KWARGS = {}      # overrides, e.g. {"max_tokens": 24000}

import pipeline
pipeline.launch_ui(PROMPT_NAME, seed=SEED, gen_kwargs=GEN_KWARGS)
```

### Run questions from the question files
`RUN_SETS`: which sets, e.g. `["obvious"]`; `None` = all three sets.
`RUN_IDS`: which questions within each set, e.g. `["Q1", "Q2"]`; `None` = all.
`SKIP_EXISTING = True` skips answers that already have a file. Set it to `False` to run them again; repeats are saved as `..._regular_prompt_2.md`, `_3`, ...

```python
RUN_SETS = None         # e.g. ["obvious", "non_obvious"]; None = every set this prompt applies to
RUN_IDS = None          # e.g. ["Q1", "Q2"]; None = every question in each set
SKIP_EXISTING = True    # False = run again even if an answer file already exists

pipeline.run_batch(RUN_SETS, RUN_IDS, prompt_names=[PROMPT_NAME], seed=SEED,
                   gen_kwargs=GEN_KWARGS, skip_existing=SKIP_EXISTING)
```
