# Detailed prompt — P1-VL-30B-A3B (Q4_K_M)

The fixed prompt lives in `prompts/prompt_detailed.md`. Type a question in the box and press **Run**.
Each answer is shown below and saved to `outputs/prompt_detailed/` as a Markdown file (plus `log.jsonl`).

```python
# One-time setup is in README.md (llama-cpp-python needs a CUDA build).
# Quick check that the GPU build is installed:
import llama_cpp; print(llama_cpp.__version__, 'GPU:', llama_cpp.llama_supports_gpu_offload())
```

```python
PROMPT_NAME = "prompt_detailed"   # the prompt text is in prompts/prompt_detailed.md
SEED = None          # set an int (e.g. 0) for reproducible sampling
GEN_KWARGS = {}      # overrides, e.g. {"max_tokens": 24000}

import pipeline
pipeline.launch_ui(PROMPT_NAME, seed=SEED, gen_kwargs=GEN_KWARGS)
```

### Run questions from `questions.json`
Set `RUN_IDS` to the questions for this run, e.g. `["Q4", "Q5", "Q6"]`, or `None` for all. Each run is saved to its own file. Both notebooks read the same `questions.json`.

```python
RUN_IDS = None   # e.g. ["Q4", "Q5", "Q6"]; None = every question in the file

import json
questions = json.load(open("questions.json"))
ids = list(questions) if RUN_IDS is None else RUN_IDS
missing = [i for i in ids if i not in questions]
assert not missing, f"Not in questions.json: {missing}"
for qid in ids:
    print(f"Running {qid}...")
    r = pipeline.ask(PROMPT_NAME, questions[qid], question_id=qid, gen_kwargs=GEN_KWARGS, seed=SEED)
    print(qid, '->', r['path'].name, '| truncated:', r['meta']['truncated'])
    print(r['answer'], '\n')
```
